```python
#!/usr/bin/env python3
from pathlib import Path
import subprocess

# --------- CONFIGURE ----------
OUTDIR = Path("demo_wavs")
START_ID = 178  # first 5-digit session id (e.g., 00178)
NUM_SESSIONS = 1  # how many consecutive session ids to create
CHUNKS_PER_SIDE = 3  # files per side (total per session = 2 * CHUNKS_PER_SIDE)
TAPE_NO = 1  # Tape number in the filename
DURATION_SEC = 2  # seconds per dummy chunk
# --------------------------------

OUTDIR.mkdir(parents=True, exist_ok=True)


def tone_freq(seq: int) -> int:
    """Simple audible ladder so you can hear the order when merged."""
    return 280 + 80 * (seq - 1)  # 280, 360, 440, ...


for sid in range(START_ID, START_ID + NUM_SESSIONS):
    session_prefix = f"CLA-{sid:05d}"

    seq = 1
    # Side A: 001..CHUNKS_PER_SIDE
    for i in range(1, CHUNKS_PER_SIDE + 1):
        fname = f"{session_prefix}_{seq:03d}_SideA_Tape{TAPE_NO}.wav"
        out = OUTDIR / fname
        subprocess.run(
            [
                "ffmpeg",
                "-hide_banner",
                "-nostdin",
                "-y",
                "-f",
                "lavfi",
                "-i",
                f"sine=frequency={tone_freq(seq)}:duration={DURATION_SEC}",
                str(out),
            ],
            check=True,
        )
        seq += 1

    # Side B: continue numbering, CHUNKS_PER_SIDE more files
    for i in range(1, CHUNKS_PER_SIDE + 1):
        fname = f"{session_prefix}_{seq:03d}_SideB_Tape{TAPE_NO}.wav"
        out = OUTDIR / fname
        subprocess.run(
            [
                "ffmpeg",
                "-hide_banner",
                "-nostdin",
                "-y",
                "-f",
                "lavfi",
                "-i",
                f"sine=frequency={tone_freq(seq)}:duration={DURATION_SEC}",
                str(out),
            ],
            check=True,
        )
        seq += 1

print(f"Created dummy WAVs in {OUTDIR.resolve()}")
```
